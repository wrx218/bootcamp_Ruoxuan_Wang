This notebook demonstrates NumPy operations, vectorization, pandas inspection, summary statistics, grouped aggregation, reusable functions, and saved outputs.


```python
from pathlib import Path
import sys
import timeit

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

candidates = [Path.cwd(), Path.cwd().parent, Path.cwd() / "homework" / "homework03"]
PROJECT_ROOT = next(
    path for path in candidates
    if (path / "data" / "raw" / "starter_data.csv").exists()
)
sys.path.insert(0, str(PROJECT_ROOT))
from src.utils import get_category_summary, get_summary_stats

PROJECT_ROOT
```

## 1. NumPy operations and vectorization


```python
numbers = np.arange(1, 11)
squared = numbers ** 2
doubled = numbers * 2

print("Original:", numbers)
print("Squared:", squared)
print("Doubled:", doubled)
```

```python
large_array = np.arange(100_000)
loop_time = timeit.timeit(lambda: [x * 2 for x in large_array], number=20)
vectorized_time = timeit.timeit(lambda: large_array * 2, number=20)

print(f"Loop time: {loop_time:.4f} seconds")
print(f"Vectorized time: {vectorized_time:.4f} seconds")
print(f"Vectorization speedup: {loop_time / vectorized_time:.1f}x")
```

## 2. Load and inspect the dataset


```python
data_path = PROJECT_ROOT / "data" / "raw" / "starter_data.csv"
df = pd.read_csv(data_path)
df.info()
```

## 3. Summary statistics and grouped aggregation


```python
summary = get_summary_stats(df)
category_summary = get_category_summary(df)

print("Numeric summary:")
print(summary)
print("\nCategory aggregation:")
print(category_summary)
```

## 4. Save processed outputs and a basic plot


```python
processed_dir = PROJECT_ROOT / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

summary.to_csv(processed_dir / "summary.csv")
category_summary.to_csv(processed_dir / "category_summary.csv")

ax = category_summary["average_sales"].plot(
    kind="bar",
    color=["#4C78A8", "#F58518", "#54A24B"],
    title="Average Sales by Category",
)
ax.set_xlabel("Category")
ax.set_ylabel("Average sales")
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.savefig(processed_dir / "category_average_sales.png", dpi=150)
plt.show()

print("Saved summary.csv, category_summary.csv, and category_average_sales.png")
```
